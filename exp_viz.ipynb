# Task 1: Exploring and Visualizing the Iris Dataset

## Objective
Learn how to load, inspect, and visualize a dataset to understand data trends and distributions.

## Dataset
Iris Dataset (loaded via seaborn library)

```python
# Import Required Libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Loading the Dataset

```python
# Load the Iris dataset
iris = sns.load_dataset('iris')
print("Dataset loaded successfully!")
```

## Inspecting the Dataset

### Basic Information

```python
# Print the shape of the dataset
print("Shape of the dataset:", iris.shape)

# Print the column names
print("\nColumn names:", list(iris.columns))

# Print the first few rows
print("\nFirst few rows:")
iris.head()
```

### Summary Statistics

```python
# Get information about the dataset
print("Dataset Info:")
iris.info()

print("\n" + "="*50 + "\n")

# Get descriptive statistics
print("Descriptive Statistics:")
iris.describe()
```

## Visualizing the Dataset

### Scatter Plot: Relationships between Features

```python
# Create a pair plot to show relationships between features
sns.pairplot(iris, hue='species', diag_kind='kde')
plt.suptitle('Scatter Plot Matrix of Iris Features', y=1.02)
plt.show()
```

### Histograms: Value Distributions

```python
# Create histograms for each feature
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for i, feature in enumerate(features):
    sns.histplot(data=iris, x=feature, hue='species', ax=axes[i], kde=True)
    axes[i].set_title(f'Distribution of {feature.replace("_", " ").title()}')

plt.tight_layout()
plt.show()
```

### Box Plots: Identifying Outliers

```python
# Create box plots for each feature to identify outliers
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for i, feature in enumerate(features):
    sns.boxplot(data=iris, x='species', y=feature, ax=axes[i])
    axes[i].set_title(f'Box Plot of {feature.replace("_", " ").title()} by Species')

plt.tight_layout()
plt.show()
```

## Conclusion

In this notebook, we successfully explored and visualized the Iris dataset. Key findings include:

- The dataset contains 150 samples with 4 numerical features and 1 categorical target (species).
- Setosa species shows clear separation from Versicolor and Virginica in petal measurements.
- There are some potential outliers, particularly in sepal width for certain species.
- The visualizations provide insights into feature distributions and relationships, which are crucial for further analysis or machine learning tasks.

This exploration demonstrates fundamental data science skills in data loading, inspection, and visualization using pandas, seaborn, and matplotlib.
